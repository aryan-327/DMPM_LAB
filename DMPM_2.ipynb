```python
# mean without using fn
l=list(map(int,input("enter the numbers: ").split()))
s=0
for i in l:
  s+=i
mean = s/len(l)
print("mean: ",mean)
```

```python
#mean using fn
import statistics
l=list(map(int,input("enter the numbers: ").split()))
mean = statistics.mean(l)
print("mean: ",mean)
```

```python
#median without fn
l= list(map(int, input('enter the nos: ').split()))

#bubble sorting
for i in range(len(l)):
  for j in range(i+1,len(l)):
    if l[i]>l[j]:
      l[i],l[j]=l[j],l[i]

#print('sorted list: ',l)
if len(l)%2==0:
  median = (l[len(l)//2-1]+l[len(l)//2])/2
else:
  median = l[len(l)//2]
print('median: ',median)
```

```python
import statistics
l=list(map(int,input("enter the numbers: ").split()))
median = statistics.median(l)
print("median: ",median)
```

```python

l = list(map(int, input("Enter the numbers: ").split()))
freq = {}
for num in l:
    freq[num] = freq.get(num, 0) + 1

max_count = 0
for count in freq.values():
    if count > max_count:
        max_count = count

modes = [key for key, count in freq.items() if count == max_count]

print("Mode(s):", modes)
```

```python
#mode with fn
import statistics
l=list(map(int,input('enter nos:').split()))
mode=statistics.mode(l)
print('mode= ',mode)
```

```python
#std and variance
import statistics
l = list(map(int, input("Enter the numbers: ").split()))
var=statistics.variance(l)
std=var**2
print('standard deviation= ',std)
print('variance= ',var)
```
