##ASSIGNMENT 4

CORRELATION OF CO-EFFICIENT  RELATION AND CO VARIANCE ANALYSIS

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Create a sample DataFrame for demonstration
data = {
    'Feature_A': np.random.rand(100) * 10,
    'Feature_B': np.random.rand(100) * 5 + np.random.rand(100) * 2,
    'Feature_C': np.random.rand(100) * 15 - np.random.rand(100) * 3
}
df = pd.DataFrame(data)

print("Sample DataFrame head:")
display(df.head())
```

```python
# Calculate the correlation matrix
correlation_matrix = df.corr()

print("Correlation Matrix:")
display(correlation_matrix)

# Visualize the correlation matrix using a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Features')
plt.show()
```

```python
# Calculate the covariance matrix
covariance_matrix = df.cov()

print("Covariance Matrix:")
display(covariance_matrix)
```

```python
# Visualize the relationship between Feature_A and Feature_B using a scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Feature_A', y='Feature_B', data=df)
plt.title('Relationship between Feature A and Feature B')
plt.xlabel('Feature A')
plt.ylabel('Feature B')
plt.grid(True)
plt.show()
```

###Assignment 6

classification using decision tree induction

```python
from sklearn.feature_extraction import DictVectorizer
from sklearn.tree import DecisionTreeClassifier

# 1. Define a custom dataset for 'price of toy' and 'buy/not buy' decision
# This is a small example dataset for demonstration purposes.
X_dict_toy = [
    {"price_of_toy": 10},
    {"price_of_toy": 20},
    {"price_of_toy": 5},
    {"price_of_toy": 25},
    {"price_of_toy": 12},
    {"price_of_toy": 30},
    {"price_of_toy": 8},
    {"price_of_toy": 18}
]

# Corresponding 'buy' or 'not buy' labels
y_toy = ["Buy", "Buy", "Buy", "Not Buy", "Buy", "Not Buy", "Buy", "Not Buy"]

# 2. Vectorize X_dict_toy to transform features into numeric vectors
vec_toy = DictVectorizer(sparse=False)
X_numeric_toy = vec_toy.fit_transform(X_dict_toy)

# 3. Induction: Train the decision tree
clf_toy = DecisionTreeClassifier(random_state=42)
clf_toy.fit(X_numeric_toy, y_toy)

print("Feature names after vectorization:", vec_toy.get_feature_names_out())

# 4. Get user input for a new toy price
try:
    user_price_input = float(input("Enter the price of the toy: "))
    user_sample_toy = [{"price_of_toy": user_price_input}]

    # Make sure to transform the new sample using the same vectorizer
    user_numeric_toy = vec_toy.transform(user_sample_toy)
    prediction_toy = clf_toy.predict(user_numeric_toy)

    print(f"\nPrediction for a toy with price {user_price_input}: **{prediction_toy[0]}**")
except ValueError:
    print("Invalid input. Please enter a numeric value for the toy price.")
```
