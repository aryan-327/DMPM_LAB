 IMPLEMENT MINMAX NORMALISATION ON A SET OF DATA POINTS

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

data = np.array([
    [10, 200],
    [25, 400],
    [50, 100],
    [75, 800],
    [100, 500]
], dtype=float)

# Initialize and transform
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(data)

print("Scikit-Learn Scaled Output:\n", scaled_data)

# Reverse back to original scale
original_data = scaler.inverse_transform(scaled_data)
```

IMPLEMENT THE MIN MAX NORMALISATION ON A DATSET

```python
import os
import kagglehub
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Download the dataset
path = kagglehub.dataset_download("himanshunakrani/iris-dataset")
print("Path to dataset files:", path)

# 2. Find and load the CSV file from the downloaded path
csv_file = [f for f in os.listdir(path) if f.endswith('.csv')][0]
df = pd.read_csv(os.path.join(path, csv_file))

print("\n--- Original Data (First 5 rows) ---")
print(df.head())

# 3. Separate numerical features from categorical columns
# (The Iris dataset typically has 4 feature columns and 1 species/label column)
feature_cols = df.select_dtypes(include=['float64', 'int64']).columns

# 4. Apply Min-Max Normalization
scaler = MinMaxScaler()
df_normalized = df.copy()
df_normalized[feature_cols] = scaler.fit_transform(df[feature_cols])

print("\n--- Normalized Data (First 5 rows) ---")
print(df_normalized.head())
```

IMPLEMENT DECIMAL SCALING NORMALISATION ON A SET OF DATA POINTS

```python
import numpy as np
import pandas as pd

def decimal_scaling_normalize(data):
    arr = np.array(data, dtype=float)
    max_abs_val = np.max(np.abs(arr))

    if max_abs_val == 0:
        return arr, 0

    # Find j such that max(|v|) / 10^j < 1
    j = int(np.ceil(np.log10(max_abs_val)))
    if max_abs_val / (10 ** j) >= 1:
        j += 1

    normalized_data = arr / (10 ** j)
    return normalized_data, j

# Sample dataset
data_points = [12, -350, 45, 980, 1200, -80]

# Normalize
normalized_points, j = decimal_scaling_normalize(data_points)

# Display results
results_df = pd.DataFrame({
    'Original Data': data_points,
    'Decimal Scaled': normalized_points
})

print(f"Scale Factor (j): {j}  (Divided by 10^{j} = {10**j})\n")
print(results_df)
```

IMPLEMENT DECIMAL SCALING NORMALISATION ON AN APPROPRIATE DATASET:

```python
import os
import kagglehub
import numpy as np
import pandas as pd

# 1. Download dataset via kagglehub
path = kagglehub.dataset_download("fedesoriano/the-boston-houseprice-data")

# 2. Locate and load the CSV file
csv_path = os.path.join(path, "boston.csv")  # Adjust filename if necessary
df = pd.read_csv(csv_path)

print("Original Data Sample:")
print(df.head())


# 3. Define Decimal Scaling Normalization function
def decimal_scaling_normalize(df_input):
    df_normalized = df_input.copy()

    for col in df_normalized.select_dtypes(include=[np.number]).columns:
        max_abs = df_normalized[col].abs().max()
        if max_abs == 0:
            continue
        # Calculate j = smallest integer such that max(|v|) / 10^j < 1
        j = int(np.ceil(np.log10(max_abs)))

        # Handle exact powers of 10 (e.g., max_abs = 100 -> log10 is 2, but 100/10^2 = 1.0)
        if max_abs / (10**j) >= 1:
            j += 1

        df_normalized[col] = df_normalized[col] / (10**j)

    return df_normalized


# 4. Apply Normalization
df_scaled = decimal_scaling_normalize(df)

print("\nNormalized Data Sample (Decimal Scaling):")
print(df_scaled.head())

# 5. Verify range constraints (all absolute values must be < 1)
print("\nMax absolute values after scaling:")
print(df_scaled.abs().max())
```

IMPLEMENT Z SCORE NORMALISATION ON 10 DATA POINTS ENTERED BY USER

```python
import numpy as np

print("Enter 10 space-separated numbers:")
while True:
    try:
        # Prompt user to enter all 10 items in a single line
        user_input = input("Numbers: ").split()
        if len(user_input) != 10:
            raise ValueError(f"Expected 10 items, but got {len(user_input)}.")

        data = np.array([float(i) for i in user_input])
        break
    except ValueError as e:
        print(f"Error: {e} Please try again.")

# Vectorized metric extraction
mean_val = np.mean(data)
std_dev = np.std(data)

print(f"\nMean (μ): {mean_val:.4f} | Standard Deviation (σ): {std_dev:.4f}\n")

# Compute all Z-scores simultaneously
if std_dev == 0:
    z_scores = np.zeros_like(data)
else:
    z_scores = (data - mean_val) / std_dev

# Display results side by side
for original, normalized in zip(data, z_scores):
    print(f"Original: {original:<8} -> Z-Score: {normalized:.4f}")
```
