write the python program to find sum and average of n numbers

```python
n = int(input('enter number of elements: '))
s=0
l=[]
print('enter elements\n')
for j in range(n):
  ele=int(input(''))
  l.append(ele)

for i in l:
  s+=i
print('sum = ',s)
avg = s/n
print('average= ',avg)
```

write a python program to count the number of vowels and consonants in a given string

```python
s= input('enter string: ')
v=c=0
for i in s:
  if i.lower() in ['a','e','i','o','u']:
    v+=1
  else:
    c+=1
print('vowels: ',v)
print('consonants : ',c)
```

factorial using function

```python
def fact(n):
  if n ==0 or n==1:
    return 1
  else:
    return n*fact(n-1)
x=int(input('enter n: '))
print('factorial= ',fact(x))
```

multiplication table of a number using fn

```python
def table(n):
  for i in range(1,11):
    print(n,'*',i,'=',n*i)
n=int(input('enter n: '))
table(n)
```

string palindrome

```python
s=input('enter string: ')
rev =s[::-1]
if s == rev:
  print('palindrome')
else:
  print('not palindrome')
```
